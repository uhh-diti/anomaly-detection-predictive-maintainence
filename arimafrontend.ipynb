```python
%%writefile app.py
import streamlit as st
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.arima.model import ARIMA
import warnings

warnings.filterwarnings("ignore")

st.title("📈 ARIMA Time Series Forecasting App")

uploaded_file = st.file_uploader("Upload your time series CSV file", type=["csv"])

if uploaded_file:
    df = pd.read_csv(uploaded_file)
    st.write("**Preview of Data**")
    st.dataframe(df.head())

    date_columns = [col for col in df.columns if 'date' in col.lower() or 'time' in col.lower()]
    numeric_columns = df.select_dtypes(include='number').columns.tolist()

    if not date_columns:
        st.error("No datetime-like column found. Please ensure your dataset has a timestamp column.")
    else:
        date_col = st.selectbox("Select the date/time column", date_columns)
        value_col = st.selectbox("Select the value column for forecasting", numeric_columns)

        if st.button("Run ARIMA Forecast"):
            df[date_col] = pd.to_datetime(df[date_col])
            df.set_index(date_col, inplace=True)
            ts = df[value_col]

            model = ARIMA(ts, order=(5, 1, 0))  # default order
            model_fit = model.fit()

            forecast = model_fit.forecast(steps=10)
            st.write("**Forecast for next 10 steps:**")
            st.write(forecast)

            fig, ax = plt.subplots()
            ts.plot(ax=ax, label="Original")
            forecast.plot(ax=ax, label="Forecast", color="red")
            plt.legend()
            st.pyplot(fig)
```
